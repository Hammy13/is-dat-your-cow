# Cow Identity Prototype Demo

This notebook demonstrates the proof-of-concept workflow:

1. Initialize the gallery from labeled side-view cow images.
2. Run image inference.
3. Run video inference with tracking.
4. Inspect the generated reports and analytics.

```python
from cow_identity_prototype.config import load_config
from cow_identity_prototype.pipeline import CowIdentityPipeline

config = load_config('cow_identity_prototype/configs/default_config.json')
pipeline = CowIdentityPipeline(config)
config.to_dict()
```

```python
gallery_path = pipeline.initialize_gallery(
    dataset_dir='cow_identity_prototype/data/gallery_train',
    gallery_path='cow_identity_prototype/artifacts/identity_gallery.json',
)
gallery_path
```

```python
image_results = pipeline.infer_images(
    image_dir='cow_identity_prototype/data/query_images',
    gallery_path='cow_identity_prototype/artifacts/identity_gallery.json',
    output_prefix='notebook_images',
)
image_results
```

```python
video_results = pipeline.infer_videos(
    video_dir='cow_identity_prototype/data/query_videos',
    gallery_path='cow_identity_prototype/artifacts/identity_gallery.json',
    output_prefix='notebook_videos',
)
video_results
```
